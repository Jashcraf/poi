```python
from poi.models import KeckTelescope
import matplotlib.pyplot as plt
from matplotlib.colors import LogNorm

kek = KeckTelescope(1.65, 20)
```

```python
kek = KeckTelescope(1.65, 20)
ref = kek.get_coronagraph_image(include_fpm=False).max()

plt.figure(figsize=[15,4])
plt.subplot(131)
plt.title("Entrance Pupil")
plt.imshow(kek.get_entrance_pupil(), cmap="gray")
plt.colorbar()
plt.subplot(132)
plt.title("Direct Image H-band")
plt.imshow(kek.get_direct_image() / ref, cmap="inferno", norm=LogNorm())
plt.colorbar()
plt.subplot(133)
plt.title("Coronagraphic Image H-band")
plt.imshow(kek.get_coronagraph_image(include_fpm=True) / ref, cmap="inferno", norm=LogNorm())
plt.colorbar(label="Normalized Intensity")
plt.show()
```
